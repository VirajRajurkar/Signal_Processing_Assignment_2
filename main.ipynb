### a. A function generating sinusoids/trains of square waves or triangle waves or saw tooth waves

Blah blah blah

```python
import numpy as np
import matplotlib.pyplot as plt

fs = 44100           # Sampling rate (samples per second)
f = 440              # Frequency (Hz)
duration = 10.0      # seconds

t = np.linspace(0, duration, int(fs*duration), endpoint=False)
sin_wave = np.sin(2 * np.pi * f * t)

plt.plot(t[:1000], sin_wave[:1000])   # plot first 1000 samples (~23 ms)
plt.title("Sine wave at 440 Hz")
plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Generating a square wave

square_wave = np.sign(np.sin(2 * np.pi * f * t))
plt.plot(t[:1000], square_wave[:1000])
plt.title("Square wave at 440 Hz")
plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Generating a triangle wave

from scipy import signal

triangle_wave = signal.sawtooth(2 * np.pi * f * t, width=0.5)
plt.plot(t[:1000], triangle_wave[:1000])
plt.title("Triangle wave (440 Hz)")
plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.show()
```

```python
# Generating a sawtooth wave

sawtooth_wave = signal.sawtooth(2 * np.pi * f * t)
plt.plot(t[:1000], sawtooth_wave[:1000])
plt.title("Sawtooth wave (440 Hz)")
plt.xlabel("Time in seconds")
plt.ylabel("Amplitude")
plt.show()
```
